# Milestone 5 — Ensembling

Combining prediction scores from multiple models can improve the final ranking of answer options.

```python
import numpy as np
```

## 1. Combine Model Scores

```python
def ensemble_scores(*model_scores, weights=None):
    if weights is None:
        weights = [1 / len(model_scores)] * len(model_scores)

    combined = np.zeros_like(model_scores[0], dtype=float)

    for scores, weight in zip(model_scores, weights):
        combined += weight * scores

    return combined
```

## 2. Rank Top-3 Options

```python
def get_top3_options(scores, options=("A", "B", "C", "D", "E")):
    ranked_indices = np.argsort(scores)[::-1][:3]
    return [options[i] for i in ranked_indices]
```

The same ranking approach can be applied to probability outputs from different models before generating the final top-3 prediction.
